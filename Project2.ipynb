```python
import pandas as pd
import matplotlib.pyplot as plt

# Dataset
data = {
    "Name": ["Alex", "John", "Sarah", "David", "Emma"],
    "Maths": [85, 70, 95, 60, 90],
    "Statistics": [90, 75, 88, 65, 92],
    "English": [80, 85, 90, 70, 88]
}

df = pd.DataFrame(data)

subjects = ["Maths", "Statistics", "English"]

# Basic information
print("STUDENT PERFORMANCE REPORT")
print("--------------------------")
print("Total students:", len(df))

# Mean
print("\nMEAN MARKS")
print(df[subjects].mean())

# Median
print("\nMEDIAN MARKS")
print(df[subjects].median())

# Standard deviation
print("\nSTANDARD DEVIATION")
print(df[subjects].std())

# Highest and lowest marks
print("\nHIGHEST MARKS")
print(df[subjects].max())

print("\nLOWEST MARKS")
print(df[subjects].min())

# Pass percentage
print("\nPASS PERCENTAGE")

for subject in subjects:
    passed = (df[subject] >= 40).sum()
    percentage = passed / len(df) * 100
    print(subject, ":", percentage, "%")

# Visualization
df.plot(x="Name", y=subjects, kind="bar")
plt.title("Student Performance Analysis")
plt.ylabel("Marks")
plt.tight_layout()
plt.show()
```
